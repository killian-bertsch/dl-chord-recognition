# Chord Recognition - Data Preprocessing

This notebook processes raw audio data and extracts features for chord recognition training.

**Input:** `data_input/` containing either:
- 00_raw folders (Ableton exports)
- 00_input files (MusicXML + WAV)

**Output:** `data_output/` containing:
- `chroma_features.csv`
- `hybrid_features.csv`
- `cqt60_features.csv`
- `cqt84_features.csv`

**Pipeline:**
1. Load raw audio files
2. Deduplicate and split bars
3. Apply pitch-shift augmentation
4. Extract audio features
5. Export to CSV for Kaggle

## 1. Setup and Imports

```python
import numpy as np
import pandas as pd
import librosa
import soundfile as sf
from pathlib import Path
import re
import warnings
from collections import defaultdict
from tqdm.notebook import tqdm

warnings.filterwarnings('ignore')

print("✓ Imports successful")
```

## 2. Configuration

```python
# Paths
RAW_DATA_DIR = Path('data_input/00_raw')  # Ableton exports
INPUT_DIR = Path('data_input/00_input')    # MusicXML + full audio
OUTPUT_DIR = Path('data_output')
OUTPUT_DIR.mkdir(exist_ok=True)

# Audio settings
SAMPLE_RATE = 22050
HOP_LENGTH = 512
TARGET_FRAMES = 87

# Augmentation
PITCH_SHIFTS = [-2, 0, 2, 4]  # Semitones

# Feature types to extract
FEATURE_TYPES = ['chroma', 'hybrid', 'cqt60', 'cqt84']

print(f"Raw data directory: {RAW_DATA_DIR}")
print(f"Input directory: {INPUT_DIR}")
print(f"Output directory: {OUTPUT_DIR}")
print(f"Pitch shifts: {PITCH_SHIFTS}")
print(f"Feature types: {FEATURE_TYPES}")
```

## 3. Chord Utilities

```python
# Chromatic scale and quality mappings
CHROMATIC_SCALE = ['C', 'Db', 'D', 'Eb', 'E', 'F', 'Gb', 'G', 'Ab', 'A', 'Bb', 'B']

QUALITY_MAP = {
    'maj7': 'maj', 'maj9': 'maj', 'maj13': 'maj', 'maj': 'maj',
    '6': 'maj', '69': 'maj',
    'min7': 'min', 'min9': 'min', 'min11': 'min', 'min6': 'min', 'min': 'min',
    'm7': 'min', 'm9': 'min', 'm11': 'min', 'm6': 'min', 'm': 'min',
    '-7': 'min', '-9': 'min', '-': 'min',
    '7': 'dom', '9': 'dom', '11': 'dom', '13': 'dom',
    '7b9': 'dom', '7#9': 'dom', '7b13': 'dom', '7#11': 'dom',
    '7sus': 'dom', '9sus': 'dom',
    'm7b5': 'hdim', 'min7b5': 'hdim', '-7b5': 'hdim',
    'ø7': 'hdim', 'ø': 'hdim',
    'dim7': 'dim', 'dim': 'dim', 'o7': 'dim', 'o': 'dim',
    'aug': 'aug', '+': 'aug',
    'sus4': 'sus', 'sus2': 'sus', 'sus': 'sus',
}

def normalize_root(root):
    """Convert sharps to flats"""
    sharp_to_flat = {'C#': 'Db', 'D#': 'Eb', 'F#': 'Gb', 'G#': 'Ab', 'A#': 'Bb'}
    return sharp_to_flat.get(root, root)

def parse_chord(chord_str):
    """Parse chord string into (root, quality) tuple"""
    chord_str = chord_str.strip()
    
    if not chord_str or chord_str.upper() in ['N.C.', 'NC', '/', '-', '']:
        return None
    
    root_match = re.match(r'^([A-G][b#]?)', chord_str)
    if not root_match:
        return None
    
    root = normalize_root(root_match.group(1))
    quality_str = chord_str[len(root_match.group(1)):].strip().replace('(', '').replace(')', '')
    
    if not quality_str:
        return (root, 'maj')
    
    for pattern, quality in QUALITY_MAP.items():
        if quality_str.startswith(pattern):
            return (root, quality)
    
    return (root, 'maj')

def transpose_note(note, semitones):
    """Transpose a note by N semitones"""
    note = normalize_root(note)
    if note not in CHROMATIC_SCALE:
        raise ValueError(f"Unknown note: {note}")
    idx = CHROMATIC_SCALE.index(note)
    new_idx = (idx + semitones) % 12
    return CHROMATIC_SCALE[new_idx]

def transpose_chord(chord_label, semitones):
    """Transpose a chord label by N semitones"""
    if ':' not in chord_label:
        return transpose_note(chord_label, semitones)
    root, quality = chord_label.split(':', 1)
    new_root = transpose_note(root, semitones)
    return f"{new_root}:{quality}"

print("✓ Chord utilities defined")
```

## 4. Audio Utilities

```python
def load_audio(audio_path, sr=SAMPLE_RATE):
    """Load audio file"""
    y, _ = librosa.load(audio_path, sr=sr, mono=True)
    return y

def pitch_shift_audio(audio, semitones, sr=SAMPLE_RATE):
    """Apply pitch shift to audio"""
    if semitones == 0:
        return audio
    return librosa.effects.pitch_shift(audio, sr=sr, n_steps=semitones)

def split_audio_in_half(audio, sr=SAMPLE_RATE):
    """Split audio into two equal halves"""
    mid_point = len(audio) // 2
    return audio[:mid_point], audio[mid_point:]

print("✓ Audio utilities defined")
```

## 5. Feature Extraction

```python
def extract_cqt84(audio, sr=SAMPLE_RATE, hop_length=HOP_LENGTH):
    """Extract 84-bin CQT features"""
    min_length = 8192
    if len(audio) < min_length:
        audio = np.pad(audio, (0, min_length - len(audio)), mode='constant')
    
    C = librosa.cqt(
        y=audio, sr=sr, hop_length=hop_length,
        n_bins=84, bins_per_octave=12,
        fmin=librosa.note_to_hz('C1')
    )
    cqt = np.abs(C)
    cqt_db = librosa.amplitude_to_db(cqt, ref=np.max)
    return cqt_db

def extract_cqt60(audio, sr=SAMPLE_RATE, hop_length=HOP_LENGTH):
    """Extract 60-bin CQT features"""
    min_length = 8192
    if len(audio) < min_length:
        audio = np.pad(audio, (0, min_length - len(audio)), mode='constant')
    
    C = librosa.cqt(
        y=audio, sr=sr, hop_length=hop_length,
        n_bins=60, bins_per_octave=12,
        fmin=librosa.note_to_hz('C1')
    )
    cqt = np.abs(C)
    cqt_db = librosa.amplitude_to_db(cqt, ref=np.max)
    return cqt_db

def extract_chroma(audio, sr=SAMPLE_RATE, hop_length=HOP_LENGTH):
    """Extract 12-bin chromagram features"""
    min_length = 8192
    if len(audio) < min_length:
        audio = np.pad(audio, (0, min_length - len(audio)), mode='constant')
    
    chroma = librosa.feature.chroma_cqt(
        y=audio, sr=sr, hop_length=hop_length, n_chroma=12
    )
    chroma_db = librosa.amplitude_to_db(chroma, ref=np.max)
    return chroma_db

def extract_hybrid(audio, sr=SAMPLE_RATE, hop_length=HOP_LENGTH):
    """Extract 36-bin hybrid features (24 LCQT + 12 Chroma)"""
    min_length = 8192
    if len(audio) < min_length:
        audio = np.pad(audio, (0, min_length - len(audio)), mode='constant')
    
    # Low-CQT (24 bins = 2 octaves)
    C = librosa.cqt(
        y=audio, sr=sr, hop_length=hop_length,
        n_bins=24, bins_per_octave=12,
        fmin=librosa.note_to_hz('C1')
    )
    lcqt = np.abs(C)
    lcqt_db = librosa.amplitude_to_db(lcqt, ref=np.max)
    
    # Chromagram (12 bins)
    chroma = librosa.feature.chroma_cqt(
        y=audio, sr=sr, hop_length=hop_length, n_chroma=12
    )
    chroma_db = librosa.amplitude_to_db(chroma, ref=np.max)
    
    # Align time dimensions
    min_frames = min(lcqt_db.shape[1], chroma_db.shape[1])
    lcqt_db = lcqt_db[:, :min_frames]
    chroma_db = chroma_db[:, :min_frames]
    
    # Stack vertically
    hybrid = np.vstack([lcqt_db, chroma_db])
    return hybrid

def normalize_features(features, target_frames=TARGET_FRAMES):
    """Normalize features to target frame count"""
    n_bins, n_frames = features.shape
    
    if n_frames < target_frames:
        pad_width = target_frames - n_frames
        features = np.pad(features, ((0, 0), (0, pad_width)), mode='constant')
    elif n_frames > target_frames:
        features = features[:, :target_frames]
    
    return features

# Feature extraction dispatcher
FEATURE_EXTRACTORS = {
    'cqt84': extract_cqt84,
    'cqt60': extract_cqt60,
    'chroma': extract_chroma,
    'hybrid': extract_hybrid
}

print("✓ Feature extraction functions defined")
```

## 6. Load and Process Raw Data

```python
def get_slice_files_deduplicated(clips_dir):
    """Get deduplicated slice files from Ableton exports"""
    clips_dir = Path(clips_dir)
    slice_files = sorted(clips_dir.glob('*.wav'))
    
    # Group by slice number
    slices_by_num = defaultdict(list)
    for f in slice_files:
        match = re.match(r'Slice (\d+)', f.stem)
        if match:
            slice_num = int(match.group(1))
            slices_by_num[slice_num].append(f)
    
    # Take the first occurrence of each slice number
    deduplicated = []
    for slice_num in sorted(slices_by_num.keys()):
        deduplicated.append((slice_num, slices_by_num[slice_num][0]))
    
    return deduplicated

def process_raw_song(song_dir):
    """Process a single song from 00_raw directory"""
    song_dir = Path(song_dir)
    song_name = song_dir.name
    
    chord_file = song_dir / f"{song_name}_chords"
    clips_dir = song_dir / f"{song_name}_clips"
    
    if not chord_file.exists() or not clips_dir.exists():
        return []
    
    # Read chords
    with open(chord_file, 'r') as f:
        chord_lines = [line.strip() for line in f if line.strip()]
    
    # Get deduplicated slices
    slice_files = get_slice_files_deduplicated(clips_dir)
    
    # Build bar structure
    bar_structure = []
    for line in chord_lines:
        if '|' in line:
            chords = [c.strip() for c in line.split('|')]
            bar_structure.append(('split', chords))
        else:
            bar_structure.append(('single', line))
    
    num_bars = len(bar_structure)
    if num_bars == 0:
        return []
    
    # Process slices
    samples = []
    
    for slice_idx_global, (slice_num, slice_path) in enumerate(slice_files):
        bar_idx = slice_idx_global % num_bars
        cycle = (slice_idx_global // num_bars) + 1
        bar_num = bar_idx + 1
        bar_type, bar_data = bar_structure[bar_idx]
        
        # Load audio
        audio = load_audio(slice_path)
        
        if bar_type == 'split':
            # Split audio and process each half
            chords = bar_data
            audio_a, audio_b = split_audio_in_half(audio)
            
            for i, (chord_str, audio_part) in enumerate(zip(chords, [audio_a, audio_b])):
                parsed = parse_chord(chord_str)
                if parsed is None:
                    continue
                
                root, quality = parsed
                label = f"{root}:{quality}"
                suffix = 'a' if i == 0 else 'b'
                
                if cycle == 1:
                    sample_id = f"{song_name}_bar{bar_num:02d}{suffix}"
                else:
                    sample_id = f"{song_name}_bar{bar_num:02d}{suffix}_cycle{cycle:02d}"
                
                samples.append({
                    'id': sample_id,
                    'audio': audio_part,
                    'label': label,
                    'song': song_name
                })
        else:
            # Single chord for whole bar
            parsed = parse_chord(bar_data)
            if parsed is None:
                continue
            
            root, quality = parsed
            label = f"{root}:{quality}"
            
            if cycle == 1:
                sample_id = f"{song_name}_bar{bar_num:02d}"
            else:
                sample_id = f"{song_name}_bar{bar_num:02d}_cycle{cycle:02d}"
            
            samples.append({
                'id': sample_id,
                'audio': audio,
                'label': label,
                'song': song_name
            })
    
    return samples

# Load all raw songs
all_samples = []

if RAW_DATA_DIR.exists():
    song_dirs = [d for d in RAW_DATA_DIR.iterdir() if d.is_dir()]
    print(f"Found {len(song_dirs)} songs in {RAW_DATA_DIR}")
    
    for song_dir in tqdm(song_dirs, desc="Loading songs"):
        samples = process_raw_song(song_dir)
        all_samples.extend(samples)
        print(f"  {song_dir.name}: {len(samples)} samples")
    
    print(f"\n✓ Loaded {len(all_samples)} total samples")
else:
    print(f"⚠ Raw data directory not found: {RAW_DATA_DIR}")
    print("Skipping raw data loading...")
```

## 7. Apply Augmentation

```python
# Apply pitch-shift augmentation
augmented_samples = []

print(f"Applying augmentation with shifts: {PITCH_SHIFTS}")

for sample in tqdm(all_samples, desc="Augmenting"):
    for shift in PITCH_SHIFTS:
        # Pitch shift audio
        shifted_audio = pitch_shift_audio(sample['audio'], shift)
        
        # Transpose label
        shifted_label = transpose_chord(sample['label'], shift)
        
        # Create new sample ID
        if shift == 0:
            new_id = sample['id']
        else:
            new_id = f"{sample['id']}_shift{shift:+d}"
        
        augmented_samples.append({
            'id': new_id,
            'audio': shifted_audio,
            'label': shifted_label,
            'song': sample['song'],
            'shift': shift
        })

print(f"\n✓ Created {len(augmented_samples)} augmented samples (from {len(all_samples)} originals)")
print(f"  Augmentation factor: {len(PITCH_SHIFTS)}x")
```

## 8. Extract Features and Export to CSV

```python
# Extract features for each feature type and save as CSV
for feature_type in FEATURE_TYPES:
    print(f"\n{'='*60}")
    print(f"Extracting {feature_type.upper()} features...")
    print(f"{'='*60}")
    
    extractor = FEATURE_EXTRACTORS[feature_type]
    
    # Extract features for all samples
    feature_data = []
    
    for sample in tqdm(augmented_samples, desc=f"Extracting {feature_type}"):
        # Extract features
        features = extractor(sample['audio'])
        features = normalize_features(features)
        
        # Flatten features (n_bins, 87) -> (n_bins * 87,)
        features_flat = features.flatten()
        
        # Create row
        row = {
            'id': sample['id'],
            'label': sample['label'],
            'root': sample['label'].split(':')[0],
            'quality': sample['label'].split(':')[1] if ':' in sample['label'] else 'maj',
            'song': sample['song'],
            'shift': sample['shift']
        }
        
        # Add feature columns
        for i, val in enumerate(features_flat):
            row[f'feature_{i}'] = val
        
        feature_data.append(row)
    
    # Create DataFrame
    df = pd.DataFrame(feature_data)
    
    # Save to CSV
    output_file = OUTPUT_DIR / f"{feature_type}_features.csv"
    df.to_csv(output_file, index=False)
    
    print(f"\n✓ Saved {feature_type} features to: {output_file}")
    print(f"  Shape: {df.shape}")
    print(f"  Columns: {len(df.columns)} (6 metadata + {len(df.columns) - 6} features)")
    print(f"  File size: {output_file.stat().st_size / (1024*1024):.2f} MB")

print(f"\n{'='*60}")
print("✓ ALL FEATURES EXTRACTED AND SAVED")
print(f"{'='*60}")
```

## 9. Summary and Verification

```python
# Display summary
print("\nPROCESSING SUMMARY")
print("="*60)
print(f"Original samples: {len(all_samples)}")
print(f"Augmented samples: {len(augmented_samples)}")
print(f"Augmentation factor: {len(PITCH_SHIFTS)}x")
print(f"Feature types: {', '.join(FEATURE_TYPES)}")
print("\nOUTPUT FILES:")

for feature_type in FEATURE_TYPES:
    output_file = OUTPUT_DIR / f"{feature_type}_features.csv"
    if output_file.exists():
        size_mb = output_file.stat().st_size / (1024*1024)
        print(f"  ✓ {output_file.name} ({size_mb:.2f} MB)")
    else:
        print(f"  ✗ {output_file.name} (not found)")

print("\n" + "="*60)
print("✓ PREPROCESSING COMPLETE!")
print("="*60)
print("\nNext steps:")
print("1. Upload CSV files to Kaggle as a dataset")
print("2. Use a separate Kaggle notebook for model training")
print("3. Train CNN models on the extracted features")
```
